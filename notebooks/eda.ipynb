```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

data_frame = pd.read_csv('../data/Netflix_Userbase.csv')

bins = [0, 25, 35, 50, 70]
labels = ['<25', '25-34', '35-49', '50+']
data_frame['age_group'] = pd.cut(data_frame['Age'], bins=bins, labels=labels)


```

```python
churn_rate = data_frame['is_churned'].mean() * 100
print(f"Overall Churn Rate: {churn_rate:.2f}%")
```

```python
churn_by_sub = data_frame.groupby('Subscription Type')['is_churned'].mean().sort_values(ascending=False) * 100

sns.barplot(x=churn_by_sub.index, y=churn_by_sub.values)
plt.title("Churn Rate by Subscription Type")
plt.ylabel("Churn Rate (%)")
plt.xlabel("Subscription Type")
plt.show()
```

```python
churn_by_device = data_frame.groupby('Device')['is_churned'].mean().sort_values(ascending=False) * 100

sns.barplot(x=churn_by_device.index, y=churn_by_device.values)
plt.title("Churn Rate by Device")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=45)
plt.show()
```

```python
churn_by_age = data_frame.groupby('age_group')['is_churned'].mean() * 100

sns.barplot(x=churn_by_age.index, y=churn_by_age.values)
plt.title("Churn Rate by Age Group")
plt.ylabel("Churn Rate (%)")
plt.xlabel("Age Group")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(data_frame[['user_lifetime_days', 'days_since_last_payment', 'is_churned']].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
